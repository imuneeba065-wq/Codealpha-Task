## World Happiness Report — Data Visualization

Visualizing the key patterns behind happiness scores across 158 countries (2015 data).

```python
import pandas as pd
df = pd.read_csv('2015.csv')
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style('whitegrid')

plt.figure(figsize=(8,5))
sns.regplot(data=df, x='Economy (GDP per Capita)', y='Happiness Score', color='teal')
plt.title('Richer Countries Tend to Be Happier', fontsize=14)
plt.xlabel('Economy (GDP per Capita)')
plt.ylabel('Happiness Score')
plt.show()
```

```python
plt.figure(figsize=(9,6))
region = df.groupby('Region')['Happiness Score'].mean().sort_values()
region.plot(kind='barh', color='coral')
plt.title('Which Regions Are Happiest?', fontsize=14)
plt.xlabel('Average Happiness Score')
plt.ylabel('')
plt.show()
```

```python
cols = ['Happiness Score','Economy (GDP per Capita)','Family','Health (Life Expectancy)','Freedom','Trust (Government Corruption)','Generosity']
plt.figure(figsize=(9,6))
sns.heatmap(df[cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('What Drives Happiness Most?', fontsize=14)
plt.show()
```

```python
top10 = df.nlargest(10, 'Happiness Score')[['Country','Happiness Score']]
bottom10 = df.nsmallest(10, 'Happiness Score')[['Country','Happiness Score']]

plt.figure(figsize=(8,6))
sns.barplot(data=top10, x='Happiness Score', y='Country', color='seagreen')
plt.title('Top 10 Happiest Countries', fontsize=14)
plt.xlabel('Happiness Score')
plt.ylabel('')
plt.show()

plt.figure(figsize=(8,6))
sns.barplot(data=bottom10, x='Happiness Score', y='Country', color='indianred')
plt.title('10 Least Happy Countries', fontsize=14)
plt.xlabel('Happiness Score')
plt.ylabel('')
plt.show()
```

## Summary

The charts above show that economic strength, family support, and health are the three factors most closely linked to happiness. Wealthier regions like Australia/New Zealand and North America rank highest, while Sub-Saharan Africa ranks lowest.
